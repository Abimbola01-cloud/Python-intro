```python
#Class Challenge
product_price = 250.00     # The product price is set to 250.
discount_rate = 0.15        # for each product we are giving a discount of 0.15.
discount_amount = product_price * discount_rate  # multiply product price and discount rate to get the discount amount.
final_price = product_price - discount_amount    # subtracting dicount amount from product price to get the final price.

print("Original price:", product_price)
print("Discount:", discount_amount)
print("Final price:", final_price)
```
