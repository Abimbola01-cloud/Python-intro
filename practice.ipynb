```python
#Project — Product Inventory Entry System
#Goal: Build an inventory entry program for a small shop.

#Your program must:

# 1) Ask for: product name, unit price, quantity in stock, and whether it is on sale (yes/no)
# 2) Store each in the correct data type (str, float, int, bool)
# 3) Calculate: total inventory value (price × quantity)
# 4) Display a formatted product card showing all fields, their types, and the total value
# 5) If the product is on sale, show a 10% discounted price as well
# 6) Use proper snake_case names and at least 4 meaningful comments

# PROJECT: Product Inventory Entry System
# Write your solution here
product = input("What product are you getting?")
unit_price = float(input("How much is the product?"))
quantity_in_stock = float(input("How many pieces are left?"))

# using comparison operators to convert answer to boolean answer
on_sale = input("Is it available on sale?")
on_sale_status = on_sale == "yes"

# displaying the formatted product card with all their fields
print("=" * 60)
print("       Product Inventory Entry System")
print("=" * 60)

print("Product Name:", product, end=".....")
print(type(product))

print("The unit price is", unit_price, end=".....")
print(type(unit_price))

print("There are",quantity_in_stock,"pieces left", end=".....")
print(type(quantity_in_stock))

print("Is it on sale?",on_sale_status, end=".....")
print(type(on_sale_status))

# calculating the total inventory value
Total_inventory_value = unit_price * quantity_in_stock
print("The total price is", round(Total_inventory_value,2))

# using IF condition to give a discount of 10% for available products. 
if on_sale_status:
    discount = unit_price * 0.1 
    print("Discount:",discount)
else:    
    print("Try again")

print("=" * 60)
print("     Thank you for using the Product Inventory System")
print("=" * 60)
```

```python
# PRACTICE QUESTION — Topic 4
# Write your solution here
# Step 1: asking for the exam score, coursework score and  attendance percentage
exam_score = float(input("enter exam score:"))
coursework = float(input("enter coursework score:"))
attendance = float(input("enter attendance percentage:"))

# Step 2: calculating the final score
final_score = (exam_score * 0.6) + (coursework * 0.4)


# Step 3: add a 5% bonus if attendance is greater than 90%
bonus_applied = True

if attendance >= 0.9:
    final_score += 0.05
    bonus_applied = True
    
# Step 4: Determine if the student passed 
# The student must have a final score of at least 50 and attendance of at least 60%
if  final_score >= 50 and attendance>= 0.6:
    result = "Pass"
else: 
    result = "Fail"

print("=" * 50)
print("         GRADE CALCULATOR")
print("=" * 50)

print("The final score is:", round(final_score,2))
print("Pass/Fail status:", result)
print("The attendance bonus applied is:", bonus_applied)

print("=" * 50)
```
